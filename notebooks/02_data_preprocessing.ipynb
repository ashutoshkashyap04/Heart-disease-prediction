```python
import pandas as pd
```

```python
# data loading

train_df = pd.read_csv("../data/raw/train.csv")
test_df = pd.read_csv("../data/raw/test.csv")

print(train_df.shape)
print(test_df.shape)
```

```python
# one hot encoding of all categorical columns

train_df = pd.get_dummies(
    train_df,
    drop_first = True
)
train_df = train_df.astype('int')

test_df = pd.get_dummies(
    test_df,
    drop_first = True
)
test_df = test_df.astype('int')
```

```python
# Standardization of columns age, restingBP, Cholesterol, MaxHR, OldPeak

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

numeric_cols = ['Age', 'RestingBP', 'Cholesterol', 'MaxHR', 'Oldpeak']
train_df[numeric_cols] = scaler.fit_transform(train_df[numeric_cols])


test_df[numeric_cols] = scaler.transform(test_df[numeric_cols])
```

```python
# saving the scaler and columns

import joblib

joblib.dump(scaler, "../models/scaler.pkl")


columns = train_df.drop("HeartDisease", axis= 1).columns.to_list()

joblib.dump(columns, "../models/columns.pkl")
```

```python
# saving the preprocessed dataset

train_df.to_csv("../data/processed/train.csv", index= False)
test_df.to_csv("../data/processed/test.csv", index = False)
```
